## Chapter 01_01_선형변환 <a class="anchor" id="chapter01_01"></a>
1. 아래와 같이 평당 금액의 데이터를 예측 오차가 가장 적은 선형 변환 모델을 찾는다.
   - (150, 200), (200, 600), (260, 500)
   - y = wx + b --> x: 입력 값, y: 출력 값, w: 가중치, b: 편향

2. 제곱 오차를 사용하여 차이가 가장 적은 선형 변환 모델을 찾는다.
    - (ỹ - y)² = err(ỹ, y)² -->  ỹ: 예측 값, y: 실제 값, err: 오차

3. 평균 가격의 오차이기 때문에 오차를 제곱하여 N개의 데이터에 대한 평균 오차를 구한다.
    - ( err(ỹ₁, y₁)² + err(ỹ₂, y₂)² + ... + err(ỹN, yN)² ) / N --> N: 데이터 개수
    - ( (ỹ₁ - y₁)² + (ỹ₂ - y₂)² + ... + (ỹN - yN)² ) / N
    - J(w,b) = ( (wx₁ + b - y₁)² + (wx₂ + b - y₂)² + ... + (wxN + b - yN)² ) / N  

4. 평균 오차(MES: Mean Error Squared)를 머신러닝 문제에대한 손실함수(Loss Function)라고 한다.

5. J(w,b)에서 w와 b를 미분하여 최소값을 찾는다.
    - 2차 함수이므로 가장 작은 값은 미분값이 0 즉 그점에서 기울기가 0인 점이다.
    - 기울기를 나타내는 도함수를 구하는데 변수가 2개 이므로 편도함수를 구한다.
    - 편도함수: 다변함수에서 하나의 변수만을 변수로 취급하고 나머지는 상수로 취급하여 미분하는 것
    - 편도함수를 나타낼대는 일반 미분기호(d/dx) 대신 ∂(라운드) 기호를 사용한다
    - ∂J/∂w = 0 --> J(w,b)가 최소화 되는 w를 구하기 위해 w에 대한 편미분을 구한다. 
    - ∂J/∂b = 0 --> J(w,b)가 최소화 되는 b를 구하기 위해 b에 대한 편미분을 구한다.

6. (150, 200), (200, 600), (260, 500)를 대입하면 손실함수는 아래와 같다.
    - J(w,b) = ( (150w + b - 200)² + (200w + b - 600)² + (260w + b - 500)² ) / 3

7. ∂J/∂w =0, ∂J/∂b = 0 연립방정식을 풀기위한 식을 유도하기 위해 합성 함수를 사용한다.
    - 합성 함수(Composite Function)란 한 함수의 출력이 다른 함수의 입력으로 사용되는 함수이다.
    - f(g(x)): g(x)의 출력이 f() 함수의 입력으로 사용된다.

8. 유도된 합성 함수
    - d₁ = 150w + b - 200 / d₂ = 200w + b - 600 / d₃ = 260w + b - 500
    - err₁ = d²₁ / err₂ = d²₂ / err₃ = d²₃
    - J = (err₁ + err₂ + err₃) / 3

9. 미분의 합 규칙과 곱 규칙을 사용한다.
    - 합 규칙: 미분 가능한 두 함수를 더한 함수의 미분은 각 함수의 미분을 더한것과 같다.
        - ∂[j₁(w) + j₂(w)] = ∂j₁(w) + ∂j₂(w)
    - 곱 규칙: 상수를 곱한 함수의 도함수는 그 함수의 도함수를 상수배 한 것과 같다.
        - ∂[c × j(w)] = c ×∂j(w)

10. (err₁ + err₂ + err₃) / 3를 미분하는 것과 err₁, err₂, err₃를 각각 미분한 후 1/3를 곱하는 것은 같다.
    - ∂J/∂w = (∂err₁/∂w + ∂err₂/∂w + ∂err₃/∂w) / 3
    - ∂J/∂b = (∂err₁/∂b + ∂err₂/∂b + ∂err₃/∂b) / 3

11. 미분 연쇄 법칙을 사용한다.
    - 합성 함수를 각 함수의 미분으로 분해하여 연쇄적으로 곱한다.
    - 체인룰이라고 부르면 복잡한 함수의 미분을 가능하게 한다.

12. d₁ = 150w + b - 200, err₁ = d²₁을 연쇄 법칙을 사용하여 미분한다.
    - 바깥 함수 미분 
        - ∂err₁/∂d₁ = 2*d²⁻¹₁ = 2d₁
    - w에 대해 안쪽 함수 미분
        - 150w + b - 200의 미분은 편 미분이기 때문에 b와 상수항은 0이 된다.
        - ∂d₁/∂w = 150
    -  b에 대해 안쪽 함수 미분
        - 150w + b - 200의 미분은 편 미분이기 때문에 w와 상수항은 0이 된다.
        - ∂d₁/∂b = 1
    - 연쇄 법칙 적용
        - ∂err₁/∂w = ∂err₁/∂d₁ * ∂d₁/∂w = 2d₁ * 150 = 300d₁ = 300(150w + b - 200) = 45000w + 300b - 60000
        - ∂err₁/∂b = ∂err₁/∂d₁ * ∂d₁/∂b = 2d₁ * 1 = 2d₁ = 2(150w + b - 200) = 300w + 2b - 400

13. d₂ = 200w + b - 600, err₂ = d²₂을 연쇄 법칙을 사용하여 미분한다.
    - 바깥 함수 미분 
        - ∂err₂/∂d₂ = 2*d²⁻¹₂ = 2d₂
    - w에 대해 안쪽 함수 미분
        - 200w + b - 600의 미분은 편 미분이기 때문에 b와 상수항은 0이 된다.
        - ∂d₂/∂w = 200
    -  b에 대해 안쪽 함수 미분
        - 200w + b - 600의 미분은 편 미분이기 때문에 w와 상수항은 0이 된다.
        - ∂d₂/∂b = 1
    - 연쇄 법칙 적용
        - ∂err₂/∂w = ∂err₂/∂d₂ * ∂d₂/∂w = 2d₂ * 200 = 400d₂ = 400(200w + b - 600) = 80000w + 400b - 240000
        - ∂err₂/∂b = ∂err₂/∂d₂ * ∂d₂/∂b = 2d₂ * 1 = 2d₂ = 2(200w + b - 600) = 400w + 2b - 1200

14. d₃ = 260w + b - 500, err₃ = d²₃을 연쇄 법칙을 사용하여 미분한다.
    - 바깥 함수 미분
        - ∂err₃/∂d₃ = 2*d²⁻¹₃ = 2d₃
    - w에 대해 안쪽 함수 미분
        - 260w + b - 500의 미분은 편 미분이기 때문에 b와 상수항은 0이 된다.
        - ∂d₃/∂w = 260
    -  b에 대해 안쪽 함수 미분
        - 260w + b - 500의 미분은 편 미분이기 때문에 w와 상수항은 0이 된다.
        - ∂d₃/∂b = 1
    - 연쇄 법칙 적용
        - ∂err₃/∂w = ∂err₃/∂d₃ * ∂d₃/∂w = 2d₃ * 260 = 520d₃ = 520(260w + b - 500) = 135200w + 520b - 260000
        - ∂err₃/∂b = ∂err₃/∂d₃ * ∂d₃/∂b = 2d₃ * 1 = 2d₃ = 2(260w + b - 500) = 520w + 2b - 1000

15. ∂J/∂w = (∂err₁/∂w + ∂err₂/∂w + ∂err₃/∂w) / 3, ∂J/∂b = (∂err₁/∂b + ∂err₂/∂b + ∂err₃/∂b) / 3를 계산한다.
    - ∂J/∂w = (45000w + 300b - 60000 + 80000w + 400b - 240000 + 135200w + 520b - 260000) / 3 = (260200w + 1220b - 560000) / 3 = 86733.33w + 406.67b - 186666.67
    - ∂J/∂b = (300w + 2b - 400 + 400w + 2b - 1200 + 520w + 2b - 1000) / 3 = (1220w + 6b - 2600) / 3 = 406.67w + 2b - 866.67

16. ∂J/∂w = 0, ∂J/∂b = 0 연립방정식을 풀어 w와 b를 구한다.
    - (1/3)(260200w + 1220b - 560000) = 0
    - (1/3)(1220w + 6b - 2600) = 0
    - w = 2.58 / b = -91.75

17.  결과로 얻은 선형모델은 아래와 같다.
    - y = 2.58x - 91.75

18. 결과로 얻은 선형모델을 사용하여 예측값을 구한다.
    - (150, 200) --> ỹ = 2.58 * 150 - 91.75 = 300.25
    - (200, 600) --> ỹ = 2.58 * 200 - 91.75 = 423.25
    - (260, 500) --> ỹ = 2.58 * 260 - 91.75 = 567.05

19. 결과로 얻은 선형모델을 사용하여 예측값과 실제값의 오차를 구한다.
    - (150, 200) --> err = 300.25 - 200 = 100.25
    - (200, 600) --> err = 423.25 - 600 = -176.75
    - (260, 500) --> err = 567.05 - 500 = 67.05

20. 결과로 얻은 선형모델을 사용하여 예측값과 실제값의 제곱 오차를 구한다.
    - (150, 200) --> err² = 100.25² = 10050.06
    - (200, 600) --> err² = -176.75² = 31245.06
    - (260, 500) --> err² = 67.05² = 4495.70

21. 결과로 얻은 선형모델을 사용하여 예측값과 실제값의 평균 제곱 오차를 구한다.
    - MES = (10050.06 + 31245.06 + 4495.70) / 3 = 15296.27
    - 제곱근은 MES = √15296.27 = 123.66

22. 모델은 평균적으로 약 123.66의 오차를 가진다. 이는 모델이 실제 데이터와 얼마나 잘 맞는지를 나타내는 지표이다.


```python
import torch
import torch.nn as nn
import torch.optim as optim
# 데이터
x_train = torch.FloatTensor([[150], [200], [260]])
y_train = torch.FloatTensor([[200], [600], [500]])
# 모델 초기화
W = torch.zeros(1, requires_grad=True)
b = torch.zeros(1, requires_grad=True)
# optimizer 설정
optimizer = optim.SGD([W, b], lr=0.00001)

nb_epochs = 10001 # 원하는만큼 경사 하강법을 반복
for epoch in range(nb_epochs + 1):

    # H(x) 계산
    hypothesis = x_train * W + b

    # cost 계산
    cost = torch.mean((hypothesis - y_train) ** 2)

    # cost로 H(x) 개선
    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    # 100번마다 로그 출력
    if epoch % 100 == 0:
        print('Epoch {:4d}/{} W: {:.3f}, b: {:.3f} Cost: {:.6f}'.format(
            epoch, nb_epochs, W.item(), b.item(), cost.item()
        ))

    print('Epoch {:4d}/{} W: {:.3f}, b: {:.3f} Cost: {:.6f}'.format(
            epoch, nb_epochs, W.item(), b.item(), cost.item()
        ))
```

```python
# 모델 초기화
W = torch.zeros(1, requires_grad=True)
b = torch.zeros(1, requires_grad=True)
# optimizer 설정
optimizer = optim.SGD([W, b], lr=0.00001)

nb_epochs = 5000000
for epoch in range(nb_epochs + 1):

    # Cost 계산
    hypothesis = x_train * W + b
    cost = torch.mean((hypothesis - y_train) ** 2)

    # cost로 H(x) 개선
    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    # 100번마다 로그 출력
    if epoch % 100 == 0:
        print('Epoch {:4d}/{} W: {:.3f}, b: {:.3f} Cost: {:.6f}'.format(
            epoch, nb_epochs, W.item(), b.item(), cost.item()
        ))
```
