## Chapter 02_01_단어 임베딩 개념 <a class="anchor" id="chapter02_01"></a>
1. 단어 임베딩은 단어를 밀집 벡터로 표현하는 방법이다.

2. 단어 임베딩은 코사인 유사도(cosine similarity)를 사용하여 단어 간의 의미적 유사성을 측정할 수 있다.
    - "영화"와 "영화관"은 코사인 유사도가 높게 나타나며, "영화"와 "음악"은 낮게 나타난다.

## Chapter 02_02_스킵그램 <a class="anchor" id="chapter02_02"></a>
1. 단어에 학습을 통해 의미를 부여하는 방법에는 CBOW(Continuous Bag of Words)와 스킵그램(Skip-gram)이 있다.

2. 주변 단어를 사용하여 중심 단어를 예측하는 방법을 스킵그램이라고한다.

3. "알랜 교수의 X 현대 컴퓨터 과학을 발전시켰다"라는 문장에서 "X"는 연구는, 논문은, 업적은 등이 될 수 있다.
    - 스킵그램은 주변의 단어를 사용하여 X를 예측하도록 학습한다.

4. "알랜 교수의 연구는 현대 컴퓨터 과학을 발전시켰다"를 스킵그램의 크기가 5로 학습
    - 중심 단어 "X"를 기준으로 좌우 2개의 단어를 주변 단어로 설정한다.
    - 첫 번째 스킵그램: 알랜 교수의 x 현대 컴퓨터 , X:연구는
    - 두 번째 스킵그램: 교수의 연구는 X 컴퓨터 과학을, X:현대
    - 세 번째 스킵그램: 연구는 현대 X 과학을 발전시켰다. X:컴퓨터

5. 첫 번째 스킵그램을 학습한다. "알랜 교수의 x 현대 컴퓨터 , X:연구는"
    - 누락된 단어를 기준으로 훈련 쌍을 생성한다.
    - [연구는, 알랜], [연구는, 교수의], [연구는, 현대], [연구는, 컴퓨터]
    - 동일한 입력 "연구는"에 대해 4개의 출력을 생성하며 이 출력은 각 단어의 확률로 나타낸다.
    - 4개 출력을 평균하여 손실을 계산하고, 역전파를 통해 가중치를 업데이트한다.
    - 이렇게 "연구는"의 임베딩 벡터가 주변 단어의 의미를 반영하도록 학습된다.



## Chapter 02_03_바이트 페어 인코딩 <a class="anchor" id="chapter02_03"></a>
1. 단어를 부분단어로 분리하여 어휘사전에 없는 단어를 처리하는 알고리즘이다.

2. "hello"라는 단어를 바이트 페어 인코딩을 통해 분리하면 "h", "e", "l", "l", "o"로 나뉘게 된다.
    - 이후 "l"과 "l"을 합쳐서 "ll"로 만들고, "e"와 "ll"을 합쳐서 "ell"로 만들고, 마지막으로 "h"와 "ell"을 합쳐서 "hell"로 만든다. 최종적으로 "hello"는 ["h", "e", "ll", "o"]로 분리된다.
    - 바이트 페어 인코딩은 이러한 과정을 반복하여 가장 자주 등장하는 부분 단어를 합쳐서 새로운 토큰을 생성한다.


## Chapter 02_04_대규모 언어모델 <a class="anchor" id="chapter02_04"></a>
1. 대규모 언어모델은 사람의 텍스트를 이해하고 생상할 수 있는 신경망 모델이다.
    
2. 다음 단어를 예측하는 방법으로 학습하며, 이를 통해 문맥을 이해하고 자연스러운 텍스트를 생성할 수 있다.

2. 대규모 언어모델의 개발은 사전훈련과 미세튜닝 단계로 이루어진다.
    - 사전훈련
        - 대규모 텍스트 데이터셋을 사용하여 모델을 학습시킨다. 
        - 이 단계에서 모델은 언어의 일반적인 패턴과 구조를 학습한다. 파운데이션 모델이라고 불리운다.
        - 사전훈련된 모델은 텍스트완성이나 퓨-샷학습 등 다양한 작업에 활용될 수 있다.
    - 미세튜닝
        - 특정 작업이나 도메인에 맞게 모델을 조정한다. 
        - 예를 들어, 감정 분석, 번역, 요약 등 특정 작업에 맞게 모델을 최적화한다.
